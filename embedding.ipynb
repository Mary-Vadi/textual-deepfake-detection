```python
X = df['text']
y = df['label']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
X_train_emb = model.encode(X_train.tolist(), show_progress_bar = True)
X_test_emb = model.encode(X_test.tolist(), show_progress_bar = True)
```

```python
from sklearn.metrics import accuracy_score

# 1. Logistic Regression
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter = 1000)
lr.fit(X_train_emb, y_train)
lr_pred = lr.predict(X_test_emb)
```

```python
# 2. SVM
from sklearn.svm import SVC
svm = SVC()
svm.fit(X_train_emb, y_train)
svm_pred = svm.predict(X_test_emb)
```

```python
# 3.Naive Bayes
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(X_train_emb, y_train)
nb_pred = nb.predict(X_test_emb)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Accuracy
print("Accuracy:", accuracy_score(y_test, lr_pred))

# Precision
print("Precision:", precision_score(y_test, lr_pred, average='weighted'))

# Recall
print("Recall:", recall_score(y_test, lr_pred, average='weighted'))

# F1 Score
print("F1 Score:", f1_score(y_test, lr_pred, average='weighted'))
```
